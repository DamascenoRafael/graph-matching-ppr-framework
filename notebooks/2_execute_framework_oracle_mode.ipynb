# README

Running the framework in *Oracle mode* is intended **solely for testing** and must not be considered a valid final result.

Providing an `oracle_order` means assigning the same predetermined ordering to the candidate vertices of both `graph_a` and `graph_b`.  
This allows you to evaluate how the framework behaves when given ideal candidate orderings.

The `oracle_order` itself may still be generated using different `SortMethods` before being passed to the framework.

```python
# importing utils notebook for helper enums and functions
%run utils.ipynb
```

```python
import sys
sys.path.append("../graph-matching-ppr-framework/")
```

```python
from framework import Framework
from models import PageRankSettings, DistanceMetricsSettings
from enums import SortMethod
from weight_matching.minimum_weight_matching_greedy import minimum_weight_matching_greedy
from utils.oracle_sorted_nodes import oracle_sorted_nodes
```

```python
# settings
page_rank_settings = PageRankSettings(is_sparse=False, max_horizon=5, return_alpha=0.05)
distance_metrics_settings = DistanceMetricsSettings(amortization_sigma=0.8, main_distance_importance_gamma=0.25)
```

```python
# graph props
graph_params = GraphParameters(
    graph_model=GraphModel.ba,
    n_nodes=512,
    edge_removal_prob=0.1
)
```

```python
instance_idx = 0
```

```python
# resolving oracle_order
original_graph = get_original_graph(graph_params, instance_idx)
oracle_order = oracle_sorted_nodes(
    original_graph=original_graph,
    sort_method=SortMethod.EGO_NET_1,
    ppr_settings=page_rank_settings
)
```

```python
graph_a, graph_b = get_input_graphs(graph_params, instance_idx)
instance = Framework(
    graph_a=graph_a,
    graph_b=graph_b,
    sort_method=SortMethod.ORACLE,
    minimum_weight_matching=minimum_weight_matching_greedy,
    page_rank_settings=page_rank_settings,
    distance_metrics_settings=distance_metrics_settings,
    oracle_order=oracle_order
)
progress = instance.execute(
    is_verbose=True
)

print(f"Final matching result with oracle_order: {progress[-1].correct_matches}/{progress[-1].matching_size}")
```
